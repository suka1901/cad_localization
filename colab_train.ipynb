# Train the CAD localisation CNN on a GPU

Runtime > Change runtime type > **T4 GPU** first. Training takes about 10-20 minutes on the GPU. After it, `cadloc run` also does the CAD edge fit and the four analyses (error budget, calibration, localisability, held-out); those run on the CPU and take roughly another 60-90 minutes on Colab.

The test captures are the same as in the CPU run, so the results compare directly with `results/summary.csv`.
